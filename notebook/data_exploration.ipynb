```python
import matplotlib.pyplot as plt

react_cols = [c for c in df_clean_NAN.columns if c.startswith("reactivity_0")]
err_cols = [c for c in df_clean_NAN.columns if c.startswith("reactivity_error_")]

moy = df_clean_NAN[react_cols].mean()
err = df_clean_NAN[err_cols].mean()

moy.index = moy.index.str.replace("reactivity_", "").astype(int)
err.index = err.index.str.replace("reactivity_error_", "").astype(int)

plt.figure(figsize=(14, 4))
plt.plot(moy.index, moy.values, color="blue", label="réactivité")
plt.fill_between(moy.index, moy - err, moy + err,
                 color="red", alpha=0.3, label="± erreur")
plt.xticks(range(0, 210, 10))
plt.xlabel("position")
plt.ylabel("réactivité")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
def plot_sequence_lengths(df):
    """Afficher la distribution des longueurs des séquences ARN."""

    # Garder une ligne par séquence
    sequences = df.drop_duplicates("sequence_id").copy()

    # Calculer la longueur de chaque séquence
    sequences["sequence_length"] = sequences["sequence"].str.len()

    # Compter les séquences pour chaque longueur
    length_counts = sequences["sequence_length"].value_counts().sort_index()

    # Tracer le graphique
    plt.figure(figsize=(12, 5))
    length_counts.plot(kind="bar")

    plt.xlabel("Longueur de la séquence (nucléotides)")
    plt.ylabel("Nombre de séquences")
    plt.title("Distribution des longueurs des séquences ARN")
    plt.tight_layout()
    plt.show()

plot_sequence_lengths(df)
```
